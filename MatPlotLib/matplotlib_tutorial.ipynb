```python
#Real data

# Real data

# Oscar Winner Movies
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

years = [2008, 2009, 2010, 2011, 2012]

oscar_revenue = [1005, 170, 427, 133, 232]  

#plot
plt.plot(years,oscar_revenue)

plt.title("Oscar movie revenue in each year")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
```

```python
# Non-Oscar Winner Movies
non_oscar_movies = ["Slumdog Millionaire", "Avatar", "Inception", "Hugo", "Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M

plt.plot(years,oscar_revenue,"o--r", linewidth=3,label="Oscar movies")
plt.plot(years,non_oscar_revenue,"<-y" ,label="Non-oscar movies")

plt.title("Oscar movies vs non-oscar movies revenue")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")

plt.legend()
```

```python
# Bar chart

movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

years = [2008, 2009, 2010, 2011, 2012]
revenue = [1005, 170, 427, 133, 232]

plt.bar(years,revenue,color="green")
plt.title("Revenue in each year for Oscar Movies")
plt.xlabel("Years")
plt.ylabel("Revenue($)")


for i in range(len(years)):
    plt.text(years[i],revenue[i]+10, str(revenue[i]),ha="center")

plt.ylim(0, max(revenue)+100)
```

```python
import numpy as np

oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
oscar_revenue = [1005, 170, 427, 133, 232]  

non_oscar_movies = ["Slumdog Millionaire", "Avatar", "Inception", "Hugo", "Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $

years = [2008, 2009, 2010, 2011, 2012]

x = np.arange(len(years))
width = 0.4

plt.bar(x-width/2,oscar_revenue,width,label="oscar movies")
plt.bar(x+width/2,non_oscar_revenue,width,label="non-oscar movies")

plt.title("Oscar Movies vs non-Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue($M)")
plt.legend()
```
