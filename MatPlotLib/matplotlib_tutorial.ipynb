```python
#Real data

# Real data

# Oscar Winner Movies
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

years = [2008, 2009, 2010, 2011, 2012]

oscar_revenue = [1005, 170, 427, 133, 232]  

#plot
plt.plot(years,oscar_revenue)

plt.title("Oscar movie revenue in each year")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
```

```python
# Non-Oscar Winner Movies
non_oscar_movies = ["Slumdog Millionaire", "Avatar", "Inception", "Hugo", "Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M

plt.plot(years,oscar_revenue,"o--r", linewidth=3,label="Oscar movies")
plt.plot(years,non_oscar_revenue,"<-y" ,label="Non-oscar movies")

plt.title("Oscar movies vs non-oscar movies revenue")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")

plt.legend()
```

```python
# Bar chart

movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

years = [2008, 2009, 2010, 2011, 2012]
revenue = [1005, 170, 427, 133, 232]

plt.bar(years,revenue,color="green")
plt.title("Revenue in each year for Oscar Movies")
plt.xlabel("Years")
plt.ylabel("Revenue($)")


for i in range(len(years)):
    plt.text(years[i],revenue[i]+10, str(revenue[i]),ha="center")

plt.ylim(0, max(revenue)+100)
```

```python
import numpy as np

oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
oscar_revenue = [1005, 170, 427, 133, 232]  

non_oscar_movies = ["Slumdog Millionaire", "Avatar", "Inception", "Hugo", "Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $

years = [2008, 2009, 2010, 2011, 2012]

x = np.arange(len(years))
width = 0.4

plt.bar(x-width/2,oscar_revenue,width,label="oscar movies")
plt.bar(x+width/2,non_oscar_revenue,width,label="non-oscar movies")

plt.title("Oscar Movies vs non-Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue($M)")
plt.legend()
```

```python
#Scatter plots

people = ["Person A", "Person B", "Person C", "Person D", "Person E",
          "Person F", "Person G", "Person H", "Person I", "Person J"]

age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
bp = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

colors = ["green" if x < 125 else "red" for x in bp]

#plt.scatter(age,bp,color=colors, s=bp, alpha=0.4)
plt.scatter(age,bp, s=bp, cmap="OrRd", c=bp )
plt.title("Age vs Blood Pressure")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")

plt.grid()
plt.colorbar(label="BP")

for i in range(len(people)):
    plt.annotate(people[i] , xy=(age[i],bp[i]) , xytext=(age[i]+1,bp[i]+1))

plt.xlim(min(age),max(age)+10)
plt.ylim(min(bp), max(bp)+5)
```

```python
# multiple datasets
legit_transactions = [
    2.99, 5.49, 8.99, 12.50, 14.99, 19.99, 23.45, 29.99, 34.99, 39.50,
    45.00, 49.99, 55.25, 60.00, 75.99, 89.99, 120.50, 150.00, 199.99,
    249.99, 300.75, 450.00, 600.00, 850.00, 1200.00
]

fraud_transactions = [
    50, 100, 150, 200, 300, 500, 500, 750, 1000, 1000,
    1200, 1500, 1500, 2000, 2500, 3000, 3000
]

#  Transaction amount distribution - Legit vs Fraud

plt.hist(fraud_transactions, edgecolor="black",bins=10,label="fraud", color="red", alpha = 0.5)
plt.hist(legit_transactions, edgecolor="black",bins=10,label="legit", color="green", alpha=0.5)

plt.title("Transaction amount distribution - Legit vs Fraud")
plt.xlabel("Amounts ($)")
plt.ylabel("Frequency")
plt.legend()
```

```python
# Multiple Datasets

import numpy as np
import matplotlib.pyplot as plt


group1 = np.random.normal(50,10,100)
group2 = np.random.normal(60,15,100)

plt.boxplot([group1, group2])
plt.xticks([1, 2], ["group1", "group2"])
plt.grid(True)
plt.show()
```

```python
#  Stack Plots

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
direct = [50, 60, 70, 80, 90, 100, 110]
organic = [30, 40, 50, 55, 60, 70, 80]
social = [20, 25, 30, 35, 40, 50, 60]

plt.stackplot(days, direct, organic, social, labels=["Direct", "Organic", "Social"])
plt.title("Marketing data for a Week")
plt.xlabel("Days")
plt.ylabel("# of customers")
plt.legend()
```
