```python
#Numpy vs Lists

#Time calculation

size = 10_000_000
py_list = list(range(size))
start = time.time()
sq_list = [x**2 for x in py_list]
end = time.time()
print(f"Python list time = {end - start} seconds")

np_array = np.array(py_list)
start = time.time()
sq_array = np_array ** 2
end = time.time()
print(f"Numpy array time = {end - start} seconds")
```

```python
#Memory
import sys
print(f"size of python lists is {sys.getsizeof(py_list) * len(py_list)} bytes")

print(f"size of numpy array is {np_array.nbytes} bytes")
```
