```python
import pandas as pd
```

```python
#Series
s = pd.Series([1,2,3,4,5])
print(s)

#index
print(s[2])
```

```python
#label
s = pd.Series([22,23,24,25] , index=["Adam","Bob","Charlie","Jack"])
print(s)
print(s["Adam"])
print(s.index)
```

```python
#vectorized
s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([10,20,30,40,50])
print(s1 + s2)
```

```python
#DataFrames
#with dictoniry of lists 
info = {
    "name":["Ram","Sparsh","Shubh"],
    "age":[19,23,21],
    "gpa":[7.0,8.3,7.8]
}

df = pd.DataFrame(info)
print(df)
print(df.index)
print(df.columns)
```

```python
# With lists of list
df = pd.DataFrame([["Adam",23],["Bob",24],["Eve",25]] , columns=["Name","Age"])
print(df)
```

```python
#With Numpy arrays
import numpy as np 

np_arr = np.array([[1,2,3],[4,5,6]])
df = pd.DataFrame(np_arr , columns=["A","B","C"])
print(df)
```

```python
#Working with CSV Data
df = pd.read_csv("employee_data.csv")
print(df)
```

```python
#AQI Dataset

df = pd.read_csv("globalAirQuality.csv")
df.describe()
```

```python
#Selecting Data
df["city"]
df[["city","aqi"]]
df.loc[0]
df.loc[0:2] #start:end , end is inclusive

df.iloc[0]
df.iloc[0:2] #start:end , end is not inclusive

#cells (row,col)
df.loc[101,"aqi"]
df.loc[0:3,["city","aqi"]]

df.columns
df.iloc[0:3 , [1,2]]

#selecting single scalar value
df.at[0,"city"]
df.iat[0,2]
```

```python
#Filtering Data
df[ df["aqi"] > 100]
df[ (df["aqi"] > 100) & (df["temperature"] > 30)]

df[ df["aqi"] > 100][["city","aqi"]] #chaining
```
