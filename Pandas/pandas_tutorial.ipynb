```python
import pandas as pd
```

```python
#Series
s = pd.Series([1,2,3,4,5])
print(s)

#index
print(s[2])
```

```python
#label
s = pd.Series([22,23,24,25] , index=["Adam","Bob","Charlie","Jack"])
print(s)
print(s["Adam"])
print(s.index)
```

```python
#vectorized
s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([10,20,30,40,50])
print(s1 + s2)
```
