```python
#AQI Dataset

df = pd.read_csv("globalAirQuality.csv")
df.describe()
```

```python
#Selecting Data
df["city"]
df[["city","aqi"]]
df.loc[0]
df.loc[0:2] #start:end , end is inclusive

df.iloc[0]
df.iloc[0:2] #start:end , end is not inclusive

#cells (row,col)
df.loc[101,"aqi"]
df.loc[0:3,["city","aqi"]]

df.columns
df.iloc[0:3 , [1,2]] #we have to specify the index of the columns too

#selecting single scalar value
df.at[0,"city"]
df.iat[0,2]
```

```python
#Filtering Data
df[ df["aqi"] > 100]
df[ (df["aqi"] > 100) & (df["temperature"] > 30)] #& , | , ~

df[ df["aqi"] > 100][["city","aqi"]] #chaining
df[df["temperature"] > 30][["city", "temperature"]]

#Query - returns copy 
df.query("aqi > 100")
df.query("aqi > 100 & temperature>30")
df.query("aqi > 100 & temperature>30")[["city","aqi"]]
#using variable
aqi_val = 100
df.query("aqi > @aqi_val")
```

```python
#Cleaning data

df = pd.read_csv("raw_data.csv")

#Handling missing values
df.isnull() 

df.isnull().sum() #count the number of missing values per column

df.dropna() #returns only not null rows
df.dropna(axis=1) #drop cols

df.fillna(0) #fill empty values with a value
#ex-
cleaned_data = df.copy()
age_mean = cleaned_data["age"].mean()

cleaned_data["age"] = cleaned_data["age"].fillna(age_mean)

df.ffill()
df.bfill()
```

```python
#Handling Duplicates
df["name"].duplicated()

df.drop_duplicates()
```

```python
#Handling Data-Types

df.dtypes

df2 = df.copy()
df2 = df2.fillna(0)

df2["age"] = df2["age"].astype("int64").copy()  #change data type 
df2.dtypes

#Handling Strings
df["gender"].str.lower()
df["gender"].str.upper()
df["name"].str.capitalize()
df["name"].str.split(" ")
```

```python
#Transforming Data - Feature Engineering

df2 = df.copy()

#apply
df2["tax"] = df2["income"].apply(lambda x : "20%" if x>60000 else "10%")

#map
gender_map = {"Male": "M" , "Female":"F", "Unknown":"U"}
df2["gender"] = df2["gender"].map(gender_map)

#assign
df2.assign(new_income = df2["income"] * 1.1)  #create a new column with a value

#replace
df2["country"] = df2["country"].replace("USA", "United States")

#rename
df2.rename(columns = {"income":"salary"}) #for row label use index instead of columns

#sort Values
df2["income"].sort_values()

df2.sort_values(["income","age"])
```
