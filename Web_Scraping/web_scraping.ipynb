```python
#Web Scraping

import requests

URL = "https://www.scrapethissite.com/pages/simple/"
res = requests.get(URL)
```

```python
import os

os.makedirs("scraped_data", exist_ok=True)

with open("scraped_data/data1.html", "w", encoding="utf-8") as f:
    f.write(res.text)
```

```python
# Beautiful Soup
from bs4 import BeautifulSoup

with open("scraped_data/data1.html", "r", encoding="utf-8") as f:
    html_content = f.read()

soup = BeautifulSoup(html_content,"lxml")
```

```python
#soup.find("h1")
all_h3 = soup.find_all("h3")
all_countries = []

for h3 in all_h3:
    name = h3.get_text(strip=True)

    population = h3.find_next("div").select("span.country-population")[0].get_text()

    all_countries.append([name,population])
```

```python
import pandas as pd 

df = pd.DataFrame(all_countries, columns=["Name","Population"])
```

```python
df.to_csv("cleaned_data/data.csv", index=False)
```

```python
import requests
from bs4 import BeautifulSoup
```

```python
page_count = 1

while True:
    URL = f"https://quotes.toscrape.com/page/{page_count}/"
    res = requests.get(URL)

    soup = BeautifulSoup(res.text , "lxml")
    quotes = soup.select("div.quote")

    if not quotes:
        print("No valid pages anymore...")
        break

    with open(f"scraped_data/quotes{page_count}.html","w", encoding="utf-8") as f:
        f.write(res.text)
        print(f"Downloaded data from page {page_count}")

    page_count = page_count + 1
        
        
```

```python
#Extract Useful Info

page_count = 1
with open(f"scraped_data/quotes{page_count}.html","r",encoding="utf-8") as f:
    html_content = f.read()

soup = BeautifulSoup(html_content,"lxml")    
```

```python
all_quotes = soup.select("div.quote")
life_quotes = []

for q in all_quotes:
    all_tags = []

    for tag in q.select(".tags .tag"):
        all_tags.append(tag.get_text())

    if "life" in all_tags:
        text = q.select_one("span.text").get_text()
        author= q.select_one("small.author").get_text()
        life_quotes.append([text,author])

print(life_quotes)
```
