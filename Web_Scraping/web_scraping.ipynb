```python
#Web Scraping

import requests

URL = "https://www.scrapethissite.com/pages/simple/"
res = requests.get(URL)
```

```python
import os

os.makedirs("scraped_data", exist_ok=True)

with open("scraped_data/data1.html", "w", encoding="utf-8") as f:
    f.write(res.text)
```

```python
# Beautiful Soup
from bs4 import BeautifulSoup

with open("scraped_data/data1.html", "r", encoding="utf-8") as f:
    html_content = f.read()

soup = BeautifulSoup(html_content,"lxml")
```

```python
#soup.find("h1")
all_h3 = soup.find_all("h3")
all_countries = []

for h3 in all_h3:
    name = h3.get_text(strip=True)

    population = h3.find_next("div").select("span.country-population")[0].get_text()

    all_countries.append([name,population])
```

```python
all_countries
```

```python
import pandas as pd 

df = pd.DataFrame(all_countries, columns=["Name","Population"])
```

```python
df.to_csv("cleaned_data/data.csv", index=False)
```
